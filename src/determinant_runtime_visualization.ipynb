```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Re-create the DataFrame
data = {
    "size": [10, 100, 1000, 10000],
    "cofactor_expansion": [1e-3, 1, 1e-4, 1e-1],  # Example values
    "lu_factorization": [1e-6, 1e-4, 1e-2, 0.1],  # Example values
    "bareiss_algo": [1e-6, 1e-5, 1e-3, 0.05]      # Example values
}

df = pd.read_csv("algo_completion_times.csv")

# Create the plot
plt.figure(figsize=(8, 6))

# Plot each column as a line
plt.plot(df["size"], df["cofactor_expansion"], label="Cofactor Expansion", marker='o')
plt.plot(df["size"], df["plu_decomp"], label="plu_decomp", marker='o')
plt.plot(df["size"], df["bareiss_algo"], label="Bareiss Algorithm", marker='o')

# Add labels and legend
plt.xlabel("Matrix Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Comparison of Determinant Calculation Methods")
plt.legend()

# Display the plot
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.show()
```
